```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
n_samples = 300

categories = ['Electronics', 'Clothing', 'Games', 'Books', 'Sport']
cat_choices = np.random.choice(categories, size=n_samples, p=[0.25, 0.25, 0.20, 0.15, 0.15])

products_dict = {
    'Electronics': ['Headphones', 'Smartwatch', 'Bluetooth Speaker', 'Powerbank', 'Monitor'],
    'Clothing': ['T-Shirt', 'Jeans', 'Jacket', 'Sneakers', 'Hoodie'],
    'Games': ['Gamepad', 'Board Game', 'Console Game', 'Gaming Mouse', 'Keyboard'],
    'Books': ['Fiction Novel', 'Sci-Fi Book', 'Python Programming', 'History Book', 'Biography'],
    'Sport': ['Yoga Mat', 'Dumbbells', 'Water Bottle', 'Bicycle Helmet', 'Fitness Band']
}
products = [np.random.choice(products_dict[cat]) for cat in cat_choices]

base_prices = {'Electronics': 120, 'Clothing': 40, 'Games': 50, 'Books': 25, 'Sport': 35}
prices = np.array([np.round(np.random.normal(base_prices[c], base_prices[c]*0.3), 2) for c in cat_choices])
prices = np.abs(prices) + 5

quantities = np.clip(np.random.poisson(lam=1.8, size=n_samples) + 1, 1, 10)
ages = np.clip(np.round(np.random.normal(loc=34, scale=10, size=n_samples)), 18, 70)
delivery_days = np.random.poisson(lam=3.2, size=n_samples) + 1
ratings = np.clip(np.round(np.random.normal(loc=4.2, scale=0.8, size=n_samples), 1), 1.0, 5.0)

df = pd.DataFrame({
    'OrderID': np.arange(1001, 1001 + n_samples),
    'Category': cat_choices,
    'Product': products,
    'Price': prices,
    'Quantity': quantities,
    'CustomerAge': ages,
    'DeliveryDays': delivery_days.astype(float),
    'Rating': ratings
})

df.loc[np.random.choice(df.index, size=10, replace=False), 'Price'] = np.nan
df.loc[np.random.choice(df.index, size=10, replace=False), 'Rating'] = np.nan
df.loc[15, 'Price'] = 4999.00      
df.loc[88, 'DeliveryDays'] = 45.0  

df.to_csv('raw_sales_data.csv', index=False)

print("Файл успішно створено і збережено у вашій папці!")
```

```python
import pandas as pd
df = pd.read_csv("raw_sales_data.csv")

print('\n\nПерші 5 рядків')
print(df.head(5))

print('\n\nОстанні 5 рядків')
print(df.tail(5))

print('\n\nВипадкова вибірка 3 рядки')
print(df.sample(3))

print('\n\nІнформація про датасет')
df.info()

print('\n\nОписова статистика')
print(df.describe())
```

```python
print('\nКількість пропущених значень у кожному стовпці:')
print(df.isna().sum())
df = df.dropna()
```

```python
print('\nКількість пропущених значень у кожному стовпці:')
print(df.isna().sum())
```

```python
df['Total'] = df['Price'] * df['Quantity']
print(df.head(5))
```

```python
total_sales = df['Total'].sum()
mean_check = df['Total'].mean()
median_check = df['Total'].median()
category_mode = df['Category'].mode()[0]
total_variance = df['Total'].var()
total_std = df['Total'].std()

print(f'Загальна сума продажів: {total_sales:,.2f}')
print(f'Середній чек: {mean_check:,.2f}')
print(f'Медіанний чек: {median_check:,.2f}')
print(f'Мода категорії товару (найчастіша): {category_mode}')
print(f'Дисперсія суми замовлення: {total_variance:,.2f}')
print(f'Стандартне відхилення: {total_std:,.2f}')
```

```python
most_expensive_order = df.loc[df['Total'].idxmax()]
cheapest_order = df.loc[df['Total'].idxmin()]

print(
    f"Найдорожче замовлення:\nID: {df['Total'].idxmax()} на суму {most_expensive_order['Total']:,.2f}"
)
print(
    f"Найдешевше замовлення:\nID: {df['Total'].idxmin()} на суму {cheapest_order['Total']:,.2f}\n"
)
```

```python
top_five = df.nlargest(5, 'Total')
print("Топ 5 найбільших замовлень: ")
print(top_five[["Product", "Price", "Quantity", "Total"]])
```
